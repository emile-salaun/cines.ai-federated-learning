# Federated learning with Flower: phase 2, the room federation

In phase 1, five simulated clients trained in your notebook. Now **you are one client**:
this notebook starts one real Flower SuperNode, which joins the room's SuperLink, run by
the trainers. The whole room trains one model together.

- Your **seat** is your share of the data. It comes from your hub user name (`user07` is seat 7).
- Each round has a **deadline**, the same rule as in step 3: a SuperNode whose local
  training would take longer is left out of that round.
- The server measures what each seat brings and shows a **leaderboard** on the projector.

Change only the number of local epochs below. No other edit is needed.

```python
from flworkshop import room
```

## 1. Pick your local epochs and join

Pick 1, 2, 5 or 10 local epochs per round. More epochs means more local work, but a
longer simulated training time: over the deadline, your update is left out. Remember
step 3: what does local work buy?

```python
room.join(epochs=2,seat=7)
```

**You should see** `Seat <n>, 2 local epoch(s)` and the `flower-supernode` command that
is now running. Your SuperNode waits until the trainers start the room run.

```python
room.status()
```

**You should see** `connected`, then, once the run has started, the number of rounds
received and updates sent going up each time you re-run the cell.

## 2. Stop and rejoin

When the trainers ask, stop your SuperNode, watch the projector, then rejoin when they
say so. You may change your local epochs when you rejoin.

```python
room.leave()
```

```python
room.join(epochs=2)
```

## 3. Find your seat on the leaderboard

The projector shows, per seat: rounds on time, late and missed, and the **impact**, how
many accuracy points the room model loses when the server leaves your update out.
